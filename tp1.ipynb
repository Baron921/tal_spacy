```python
import xml.etree.ElementTree as ET

def charger_corpus_xml(chemin_fichier):
    """
    Extrait le texte des balises <p> à l'intérieur des balises <texte>
    pour chaque document <doc> du fichier XML.
    """
    corpus = []
    # Analyse du fichier XML
    arbre = ET.parse(chemin_fichier)
    racine = arbre.getroot()

    # Parcours de chaque document dans le corpus
    for doc in racine.findall('.//doc'):
        texte_complet = []
        # Extraction de tous les paragraphes
        for p in doc.findall('.//texte/p'):
            if p.text:
                texte_complet.append(p.text.strip())

        # Joindre les paragraphes pour former le texte du document
        corpus.append(" ".join(texte_complet))

    return corpus

# Chargement des corpus (à adapter si les fichiers sont dans un sous-dossier)
fichier_en = "deft09_parlement_appr_en.xml"
fichier_fr = "deft09_parlement_appr_fr.xml"

corpus_en = charger_corpus_xml(fichier_en)
corpus_fr = charger_corpus_xml(fichier_fr)

print(f"Chargement terminé : {len(corpus_en)} documents Anglais, {len(corpus_fr)} documents Français.")

# Aperçu du premier document français pour vérification
if corpus_fr:
    print("\n--- Aperçu du premier document (FR) ---")
    print(corpus_fr[0][:200] + "...")
```
